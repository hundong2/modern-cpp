# 02-flow: 조건문, 반복문, 함수

목표는 조건과 반복을 작은 함수로 묶고, 값 전달과 참조 전달의 차이를 직접 확인하는 것입니다.

## 조건문

`if`, `else if`, `else`는 위에서 아래로 검사합니다. 먼저 만족한 분기만 실행됩니다.

```c++
#include <iostream>
#include <string>

{
    auto grade_of = [](int score) {
        if (score >= 90) return std::string{"A"};
        if (score >= 80) return std::string{"B"};
        if (score >= 70) return std::string{"C"};
        if (score >= 60) return std::string{"D"};
        return std::string{"F"};
    };

    std::cout << "85 -> " << grade_of(85) << "\n";
    std::cout << "59 -> " << grade_of(59) << "\n";
}
```

## 반복문

컨테이너를 읽기만 할 때는 `const auto&`를 기본 선택지로 둡니다. 작은 정수처럼 복사가 싼 값은 `int value`처럼 값으로 받아도 괜찮습니다.

```c++
#include <iostream>
#include <vector>

{
    std::vector<int> values = {1, 2, 3, 4, 5, 6};
    int even_sum = 0;

    for (int value : values) {
        if (value % 2 == 0) {
            even_sum += value;
        }
    }

    std::cout << "sum of evens: " << even_sum << "\n";
}
```

## 값 전달과 참조 전달

값 전달은 복사본을 바꿉니다. 참조 전달은 원본을 바꿉니다. 함수가 값을 수정해야 할 때만 비-const 참조를 사용하세요.

```c++
#include <iostream>

{
    auto by_value = [](int x) { x = 99; };
    auto by_ref = [](int& x) { x = 99; };

    int a = 1;
    by_value(a);
    std::cout << "after by_value: " << a << "\n";

    by_ref(a);
    std::cout << "after by_ref  : " << a << "\n";
}
```

## 실습: 점수 보정과 FizzBuzz

`normalize_score`의 기준을 바꾸거나 FizzBuzz 범위를 늘려 보세요.

```c++
#include <iostream>

{
    auto normalize_score = [](int& score) {
        if (score < 0) score = 0;
        else if (score > 100) score = 100;
    };

    int score = 107;
    normalize_score(score);
    std::cout << "normalized score: " << score << "\n";

    for (int i = 1; i <= 15; ++i) {
        if (i % 15 == 0) std::cout << "FizzBuzz ";
        else if (i % 3 == 0) std::cout << "Fizz ";
        else if (i % 5 == 0) std::cout << "Buzz ";
        else std::cout << i << ' ';
    }
    std::cout << "\n";
}
```

## 체크포인트

- 분기가 짧아도 조건 순서를 먼저 검토한다.
- 반복문 내부에서 바뀌는 상태를 최소화한다.
- 함수 인자가 원본을 바꾸지 않으면 값 또는 `const&`로 받는다.
- 원본 변경이 목적일 때만 `T&`를 사용한다.
