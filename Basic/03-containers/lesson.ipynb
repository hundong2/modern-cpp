# 03-containers: string, vector, algorithm

목표는 표준 컨테이너를 직접 다루고 `<algorithm>`과 lambda로 반복 코드를 줄이는 것입니다.

## `std::string`

`std::string`은 문자열 길이, 연결, 부분 문자열 탐색 같은 기본 기능을 제공합니다.

```c++
#include <iostream>
#include <string>

{
    std::string text = "modern cpp";
    text += " practice";

    std::cout << text << "\n";
    std::cout << "size: " << text.size() << "\n";
    std::cout << "has cpp: " << std::boolalpha << (text.find("cpp") != std::string::npos) << "\n";
}
```

## `std::vector`

`std::vector`는 가장 자주 쓰는 동적 배열입니다. 원소 추가, 순회, 크기 확인이 쉽습니다.

```c++
#include <iostream>
#include <vector>

{
    std::vector<int> scores = {80, 95, 70};
    scores.push_back(88);

    int sum = 0;
    for (int score : scores) {
        sum += score;
    }

    std::cout << "count: " << scores.size() << "\n";
    std::cout << "average: " << static_cast<double>(sum) / scores.size() << "\n";
}
```

## `<algorithm>`과 lambda

정렬, 개수 세기, 제거 조건처럼 반복 패턴이 뚜렷한 코드는 표준 알고리즘으로 표현합니다.

```c++
#include <algorithm>
#include <iostream>
#include <string>
#include <vector>

{
    std::vector<std::string> words = {"cpp", "modern", "lambda", "stl", "vector"};

    std::sort(words.begin(), words.end());
    auto long_count = std::count_if(words.begin(), words.end(), [](const std::string& word) {
        return word.size() >= 5;
    });

    std::cout << "sorted: ";
    for (const auto& word : words) {
        std::cout << word << ' ';
    }
    std::cout << "\nlong words: " << long_count << "\n";
}
```

## 실습: 단어 빈도 세기

`words`에 단어를 더 넣고 가장 많이 나온 단어를 찾아보세요.

```c++
#include <iostream>
#include <string>
#include <unordered_map>
#include <vector>

{
    std::vector<std::string> words = {"cpp", "modern", "cpp", "vector", "lambda", "cpp"};
    std::unordered_map<std::string, int> counts;

    for (const auto& word : words) {
        ++counts[word];
    }

    for (const auto& [word, count] : counts) {
        std::cout << word << ": " << count << "\n";
    }
}
```

## 체크포인트

- `std::vector<T>`는 크기가 변하는 배열이다.
- 읽기 전용 순회는 `const auto&`를 기본으로 둔다.
- 정렬, 필터링, 카운팅은 직접 반복문을 쓰기 전에 `<algorithm>`을 먼저 찾는다.
- lambda는 짧은 조건 또는 동작을 그 자리에서 표현할 때 유용하다.
