## 기본 타입

C++은 정적 타입 언어입니다. 변수의 타입은 컴파일 시점에 정해지고, 타입에 맞지 않는 연산은 컴파일 오류가 됩니다.

```c++
#include <iostream>
#include <string>

{
    int age = 30;
    double ratio = 3.14;
    char grade = 'A';
    bool passed = true;
    std::string name = "kim";

    std::cout << std::boolalpha;
    std::cout << name << ", " << age << ", " << ratio << ", " << grade << ", " << passed << "\n";
}
```

## `auto`, `const`, `constexpr`

`auto`는 동적 타입이 아니라 컴파일러가 초기값을 보고 타입을 정하는 문법입니다. 값이 바뀌면 안 되면 `const`, 컴파일 시점 상수로 쓸 수 있으면 `constexpr`를 사용합니다.

```c++
#include <iostream>
#include <vector>

{
    auto count = 4;                  // int
    const auto unit_price = 1200.0;  // double, 이후 변경 불가
    constexpr int tax_percent = 10;  // 컴파일 시점 상수
    std::vector<int> scores = {80, 90, 75};

    std::cout << "count type is decided at compile time\n";
    std::cout << "total before tax: " << count * unit_price << "\n";
    std::cout << "tax percent: " << tax_percent << "\n";
    std::cout << "score count: " << scores.size() << "\n";
}
```

## 실습: 영수증 계산

아래 값들을 바꿔 실행해 보세요. `subtotal`, `discount`, `total`의 타입을 명시하지 않아도 `auto`가 적절한 타입을 추론합니다.

```c++
#include <iomanip>
#include <iostream>
#include <string>

{
    std::string item = "notebook";
    int quantity = 3;
    double unit_price = 12500.0;
    double discount_rate = 0.10;

    // TODO: 위 네 값을 바꾸고 출력 결과를 비교해 보세요.
    auto subtotal = quantity * unit_price;
    auto discount = subtotal * discount_rate;
    auto total = subtotal - discount;

    std::cout << std::fixed << std::setprecision(0);
    std::cout << "item     : " << item << "\n";
    std::cout << "subtotal : " << subtotal << "\n";
    std::cout << "discount : " << discount << "\n";
    std::cout << "total    : " << total << "\n";
}
```

## 체크포인트

- `std::cout << value << '\n'` 형태로 출력한다.
- `char`는 작은따옴표, 문자열은 큰따옴표를 사용한다.
- `auto`는 타입을 숨기는 도구가 아니라 중복을 줄이는 도구다.
- 바뀌지 않아야 하는 값은 기본적으로 `const`로 둔다.
