## 1. 컴파일 언어란 무엇인가

Python은 인터프리터가 `.py` 를 한 줄씩 읽어 실행한다. C++은 그렇지 않다:

```
your_code.cpp  ──[전처리기]──▶  확장된 .cpp
                                    │
                                    ▼
                              [컴파일러]   ──▶  .o (오브젝트 파일, 기계어 + 미해결 심볼)
                                    │
                                    ▼
                              [링커]      ──▶  실행파일 (a.out)
```

각 단계가 무슨 일을 하는지:

- **전처리기 (preprocessor)** — `#include`, `#define`, `#ifdef` 같은 `#`으로 시작하는 지시어를 처리한다. `#include <iostream>`은 *iostream 파일의 내용을 그 자리에 그대로 갖다 붙인다*. 이게 끝이다. 마법은 없다.
- **컴파일러** — C++ 코드를 그 파일에 한해 기계어로 바꾼다. 다른 파일에서 정의된 함수가 있으면 "이 이름의 심볼이 어디선가 정의되어 있을 것"이라고 표시만 해두고 넘어간다.
- **링커** — 여러 `.o` 파일과 라이브러리를 조립해 최종 실행파일을 만든다. "어디선가 정의되어 있을 것"이라 표시된 심볼을 실제로 찾아 연결한다.

이 3단계를 알면 나중에 보게 될 무수한 빌드 에러를 두 부류로 나눌 수 있게 된다 — *컴파일 에러* (한 파일 안에서 문법이 틀림) vs *링커 에러* (`undefined reference to ...`, 정의를 못 찾음).


## 2. 첫 코드 — Hello World

xeus-cling 셀에서는 이렇게 쓴다 (`main` 없이):

````c++
#include <iostream>

std::cout << "Hello, C++!" << std::endl;
````

읽어볼 것:

- `#include <iostream>` — 입출력 스트림 라이브러리를 끌어온다. `< >` 는 시스템(표준) 헤더, `" "` 는 내 프로젝트 헤더라는 관습이 있다.
- `std::cout` — std 네임스페이스에 있는 `cout`. `cout`은 'character output stream'의 줄임말이고, 표준 출력(터미널)에 연결된 객체다.
- `<<` — 여기서는 산술 시프트가 아니라 **스트림 삽입 연산자**다. `cout`에 데이터를 흘려넣는다고 읽으면 된다. 같은 기호가 문맥에 따라 다른 일을 하는 게 C++의 첫 충격이다 (이걸 **연산자 오버로딩**이라 부른다, Lesson 04에서 자세히).
- `std::endl` — 줄바꿈 문자 `'\n'`을 출력하고 추가로 출력 버퍼를 비운다. 사실 대부분 그냥 `"\n"`이 더 빠르다. `endl`은 "꼭 지금 화면에 찍혀야 한다"는 의미가 있을 때만 써라.

## 3. `using namespace std;` 를 써야 하나?

학생용 책을 보면 `using namespace std;` 가 흔히 등장한다. 이걸 쓰면 `std::` 를 매번 안 붙여도 된다. **그러나 헤더 파일에서는 절대 쓰지 말 것.** 그 헤더를 포함하는 모든 파일에 std 네임스페이스가 풀려 이름 충돌의 원인이 된다. 학습 단계에서는 `std::`를 명시적으로 붙이는 습관을 권한다 — 어느 라이브러리에서 온 이름인지 코드만 봐도 보이기 때문이다.

## 4. 변수, 타입, 그리고 `auto`

C++은 **정적 타입** 언어다. 변수에 타입이 박혀 있고 컴파일 타임에 검사된다.

````c++
#include <iostream>
#include <string>

int    age   = 30;
double pi    = 3.14159;
char   grade = 'A';
bool   ok    = true;
std::string name = "Alice";

std::cout << name << ", " << age << ", " << pi << ", " << grade << ", " << ok << std::endl;
````

주의할 점들:

- `'A'` (작은따옴표) 는 `char` 한 글자, `"A"` (큰따옴표) 는 길이 1인 C 문자열(`const char*`). 둘은 완전히 다른 타입이다.
- `bool` 을 `cout` 으로 출력하면 1/0 으로 찍힌다. `std::boolalpha` 를 흘려넣으면 `true`/`false` 로 바뀐다.
- `int` 의 크기는 **구현 정의**다. 거의 항상 4바이트지만 표준이 보장하는 건 "최소 16비트 이상". 정확한 크기가 필요하면 `<cstdint>` 의 `int32_t`, `int64_t` 같은 고정폭 타입을 써라.

### `auto` — 타입 추론

C++11부터 컴파일러가 우변을 보고 타입을 추론하게 할 수 있다:

````c++
#include <iostream>
#include <vector>

auto x = 42;             // int
auto y = 3.14;           // double
auto v = std::vector<int>{1, 2, 3};  // std::vector<int>

std::cout << x << " " << y << " " << v.size() << std::endl;
````

`auto` 는 동적 타입이 아니다. **컴파일 타임에 결정된 정적 타입**일 뿐이다. 일단 결정된 후엔 절대 안 바뀐다.

`auto` 를 쓰는 가장 좋은 경우는 타입 이름이 길거나(이터레이터, 람다 반환 타입) 복잡한 템플릿 인스턴스화일 때다. 단순한 `int x = 42;` 까지 `auto x = 42;` 로 쓸 필요는 없다 — 가독성을 해친다.

## 5. 제어 흐름 — 다른 언어와 다른 점만

`if`, `for`, `while`, `switch` 는 다 있다. 흥미로운 건:

### 범위 기반 for (C++11)

````c++
#include <iostream>
#include <vector>

std::vector<int> nums = {10, 20, 30, 40};

for (auto n : nums) {           // 값 복사
    std::cout << n << " ";
}
std::cout << "\n";

for (auto& n : nums) {          // 참조 (수정 가능)
    n *= 2;
}

for (const auto& n : nums) {    // const 참조 (읽기 전용, 복사 없음)
    std::cout << n << " ";
}
std::cout << "\n";
````

**규칙으로 외워둘 것**: 컨테이너를 순회할 때

- 값을 안 바꾸고 작은 타입(int, double 등)이면 `auto`
- 값을 바꾸려면 `auto&`
- 값을 안 바꾸지만 큰 객체(std::string, struct 등)이면 `const auto&` — 매 순회마다의 복사를 피한다

이 차이를 모르면 나중에 `for (auto x : huge_vector_of_strings)` 같은 코드로 성능을 50배쯤 잃을 수 있다.

## 6. 함수 — 선언과 정의

````c++
#include <iostream>

// 함수 정의 (선언 + 본문)
int add(int a, int b) {
    return a + b;
}

std::cout << add(3, 4) << std::endl;
````

실제 프로젝트에서는 이렇게 분리한다:

```cpp
// math.h  — 헤더 (선언)
int add(int a, int b);

// math.cpp — 소스 (정의)
#include "math.h"
int add(int a, int b) { return a + b; }

// main.cpp
#include "math.h"
int main() { return add(3, 4); }
```

왜 분리하나? 다른 파일이 `add`를 쓰려면 *시그니처만* 알면 된다 (컴파일러는 선언만 봐도 호출 코드를 만들 수 있다). 정의는 링커가 합쳐줄 때 한 번만 있으면 된다. 같은 함수 정의가 여러 `.cpp`에 들어가면 **One Definition Rule (ODR)** 위반으로 링커 에러가 난다. 이게 헤더에 함수 본문을 막 쓰면 안 되는 이유다 (`inline`이나 템플릿이면 예외 — 나중에).

## 7. 참조와 포인터 — 미리 보는 한 컷

자세한 건 Lesson 02에서 다루지만, 함수 인자에서 자주 보게 될 두 가지를 미리:

````c++
#include <iostream>

void by_value(int x)    { x = 99; }   // 복사. 원본 안 바뀜
void by_ref(int& x)     { x = 99; }   // 참조. 원본 바뀜
void by_const_ref(const int& x) { std::cout << x << "\n"; }  // 읽기만 (큰 객체에 적합)

int a = 1;
by_value(a);     std::cout << "after by_value: "    << a << "\n";
by_ref(a);       std::cout << "after by_ref: "      << a << "\n";
by_const_ref(a); std::cout << "after by_const_ref: " << a << "\n";
````

## 9. 정리

이 강의에서 익혀야 할 것:

- C++은 컴파일 → 링크 단계를 거친다. 에러는 두 단계 중 어디서 났는지를 먼저 본다.
- `std::` 를 명시적으로 붙여라. 어느 라이브러리에서 왔는지 보인다.
- 변수 타입은 컴파일 타임에 고정된다. `auto` 는 추론일 뿐 동적 타입이 아니다.
- 컨테이너 순회는 `const auto&` 를 기본값처럼 써라.
- 헤더 = 선언, 소스 = 정의. ODR 한 번 위반하면 링커가 화를 낸다.
